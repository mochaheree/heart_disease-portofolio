# **Initial Library**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Load CSV**

```python
#Load CSV
df = pd.read_csv("/content/heart_disease.csv")
df.head()
```

```python
df.info()
```

# **Cleaning & Preparation Data**

```python
# Summarize Missing Value
missing_values = df.isna().sum().sort_values(ascending=False)
pd.DataFrame({"Missing_count": missing_values})

## there are some columns that have missing values
```

```python
# Check for Duplicate

dup_count = df.duplicated().sum()
print("Duplicate rows:", dup_count)

## if there is no duplicate data, then skip this
```

```python
# Standardize Column Names
print("Before :")
df.columns.tolist()
```

```python

# to make it easier to initialize variables, I will tidy up the column names

df.columns = (
    df.columns
    .str.replace(" ", "_") #replace spaces with "_"
    .str.lower() #change to lowercase
)

print("After :")
df.columns
```

```python
# Tidy up catagorical columns

cat_col = df.select_dtypes(include="object").columns.tolist()
cat_col
```

```python
for c in cat_col:
  df[c] = df[c].astype(str).str.strip().str.title()

df.replace(["Nan", "Na", "None", "NaN", ""], np.nan, inplace=True)

df.info()
```

```python
target_col = "heart_disease_status"
df[target_col].value_counts(dropna=False)
```

```python
# Separate numeric vs categorical

num_cols = df.select_dtypes(include="number").columns.tolist()
cat_cols = df.select_dtypes(include="object").columns.tolist()

num_cols, cat_cols
```

```python
# Fill missing value by numeric & categorical

for c in num_cols:
  med = df[c].median()
  df[c] = df[c].fillna(med)

for c in cat_cols:
  if c == target_col:
    continue
  mode_val = df[c].mode(dropna=True)
  if len(mode_val) > 0:
    df[c] = df[c].fillna(mode_val[0])
  else:
    df[c] = df[c].fillna("Unknown")

# Numeric missing values were imputed using MEDIAN to reduce outlier bias.
# Categorical missing values were filled with MODE to preserve dominant distribution.
# The target label was not imputed
```

```python
pd.DataFrame({
    "dtype": df.dtypes,
    "missing": df.isna().sum()
}).head(21)
```

```python
df["age"].min()
```

```python
# Feature Engineering for EDA & dashboard

if "age" in df.columns:
  bins = [0, 29, 39, 49, 59, 69, 120]
  labels = ["18-29", "30-39", "40-49", "50-59", "60-69", "70 >"]
  df["age_group"] = pd.cut(df["age"], bins=bins, labels=labels, right=True)

df["age_group"].value_counts()
```

```python
df.columns.tolist()
```

```python
# Simple lifestyle risk score

risk_score = pd.Series(0, index=df.index)

if "smoking" in df.columns:
  risk_score += (df["smoking"] == "Yes").astype(int)

if "alcohol_consumption" in df.columns:
  risk_score += (df["alcohol_consumption"].isin(["High", "Yes"])).astype(int)

if "exercise_habits" in df.columns:
  risk_score += (df["exercise_habits"].isin(["Low", "None"])).astype(int)

if "stress_level" in df.columns:
  risk_score += (df["diabetes"] == "Yes").astype(int)

if "sleep_hours" in df.columns:
  risk_score += (df["sleep_hours"] < 6).astype(int)

if "sugar_consumption" in df.columns:
  risk_score += (df["sugar_consumption"] == "High").astype(int)

df["risk_score_light"] = risk_score

df["risk_segment"] = pd.cut(
    df["risk_score_light"],
    bins=[-1, 1, 3, 10],
    labels=["Low", "Medium", "High"],
    right=True
)

df[["risk_score_light", "risk_segment"]].head()
```
